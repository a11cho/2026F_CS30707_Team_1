기본 FULL_TRAINING=False는 짧은 연결 검사입니다. True로 바꾸면 해당 알고리즘 전체 12후보 학습이 시작됩니다.

```python
import sys
sys.dont_write_bytecode = True
from pathlib import Path
root = Path.cwd()
if root.name == "smoke_test_mc_td":
    root = root.parent
if not (root / "territory_rl").is_dir():
    raise RuntimeError("프로젝트 루트 또는 smoke_test_mc_td 폴더에서 노트북을 여세요.")
sys.path.insert(0, str(root))
print(sys.executable)
```

```python
from dataclasses import replace
from smoke_test_mc_td.config import agreed_config
from smoke_test_mc_td.run import run_grid

FULL_TRAINING = False
cfg = replace(agreed_config(), algorithms=("mc",))
if not FULL_TRAINING:
    # 최신 네 조건과 세 seed는 유지하고 연결 검사용 예산만 줄입니다.
    cfg = replace(cfg, episodes=4, ticks=12, validation_pairs=1, test_pairs=1,
                  diagnostic_interval=0, eval_opponents=("random",))
print(cfg)
output = run_grid(cfg)
```

```python
import json
results = json.loads((output / "results.json").read_text(encoding="utf-8"))
for item in results["selected"]:
    print(item["condition"], "seed", item["seed"], "validation", item["validation"]["score"], "test", item["test"]["score"])
```
